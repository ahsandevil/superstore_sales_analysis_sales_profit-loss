```python
import pandas as pd
df = pd.read_csv("amazaon_sales_analysis/superstore.csv", encoding="latin1")
df.head()
```

```python
print(df.isna().sum())
```

```python
print(df.duplicated().any())
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.head(10)
```

```python
print("Data types:")
print(df.dtypes)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])

#extract order day,month,year
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
print(df.dtypes)
```

```python
df.head(5)
```

```python
import matplotlib.pyplot as plt
region_analysis = df.groupby('region')[['sales', 'profit']].sum().reset_index()


region_analysis = region_analysis.sort_values('sales', ascending=False)

print(region_analysis)

#bar chart used for different visulization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))


axes[0].bar(region_analysis['region'], region_analysis['sales'],
            color="blue")
axes[0].set_title('total sales by region', fontsize=14)
axes[0].set_xlabel('Region')
axes[0].set_ylabel('Total Sales ($)')

# Profit by Region
axes[1].bar(region_analysis['region'], region_analysis['profit'],
            color="yellow")
axes[1].set_title('Total Profit by Region', fontsize=14)
axes[1].set_xlabel('Region')
axes[1].set_ylabel('Total Profit ($)')

plt.tight_layout()
plt.show()
```

```python
country_analysis = df.groupby('country')[['sales', 'profit']].sum().reset_index()

country_analysis = country_analysis.sort_values('sales', ascending=False)

print(country_analysis)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sales by Country
axes[0].bar(country_analysis['country'], country_analysis['sales'], color="blue")
axes[0].set_title('Total Sales by Country', fontsize=14)
axes[0].set_xlabel('Country')
axes[0].set_ylabel('Total Sales ($)')
axes[0].tick_params(axis='x', rotation=45)

# Profit by Country
axes[1].bar(country_analysis['country'], country_analysis['profit'], color="yellow")
axes[1].set_title('Total Profit by Country', fontsize=14)
axes[1].set_xlabel('Country')
axes[1].set_ylabel('Total Profit ($)')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
segment_analysis = df.groupby('segment')[['sales', 'profit']].sum().reset_index()

segment_analysis = segment_analysis.sort_values('sales', ascending=False)

print(segment_analysis)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sales by Country
axes[0].bar(segment_analysis['segment'], segment_analysis['sales'], color="blue")
axes[0].set_title('Total Sales by segment', fontsize=14)
axes[0].set_xlabel('segment')
axes[0].set_ylabel('Total Sales ($)')
axes[0].tick_params(axis='x', rotation=45)

# Profit by Country
axes[1].bar(segment_analysis['segment'], segment_analysis['profit'], color="yellow")
axes[1].set_title('Total Profit by segment', fontsize=14)
axes[1].set_xlabel('segment')
axes[1].set_ylabel('Total Profit ($)')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
shipping_analysis = df.groupby('ship_mode')[['sales', 'profit']].sum().reset_index()
shipping_analysis = shipping_analysis.sort_values('sales', ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(shipping_analysis['ship_mode'], shipping_analysis['sales'], color="blue")
axes[0].set_title('Total Sales by Shipping Mode')
axes[0].set_xlabel('Shipping Mode')
axes[0].set_ylabel('Sales')
axes[0].tick_params(axis='x', rotation=45)

axes[1].bar(shipping_analysis['ship_mode'], shipping_analysis['profit'], color="green")
axes[1].set_title('Total Profit by Shipping Mode')
axes[1].set_xlabel('Shipping Mode')
axes[1].set_ylabel('Profit')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
city_analysis = df.groupby('city')[['sales', 'profit']].sum().reset_index()
city_analysis = city_analysis.sort_values('sales', ascending=False).head(10)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(city_analysis['city'], city_analysis['sales'], color="blue")
axes[0].set_title('Top 10 Cities by Sales')
axes[0].set_xlabel('City')
axes[0].set_ylabel('Sales')
axes[0].tick_params(axis='x', rotation=45)

axes[1].bar(city_analysis['city'], city_analysis['profit'], color="green")
axes[1].set_title('Top 10 Cities by Profit')
axes[1].set_xlabel('City')
axes[1].set_ylabel('Profit')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
category_analysis = df.groupby('category')[['sales', 'profit']].sum().reset_index()
category_analysis = category_analysis.sort_values('sales', ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(category_analysis['category'], category_analysis['sales'], color="blue")
axes[0].set_title('Total Sales by Category')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Sales')

axes[1].bar(category_analysis['category'], category_analysis['profit'], color="green")
axes[1].set_title('Total Profit by Category')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Profit')

plt.tight_layout()
plt.show()
```

```python
subcat_analysis = df.groupby('sub-category')[['sales', 'profit']].sum().reset_index()
subcat_analysis = subcat_analysis.sort_values('sales', ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].bar(subcat_analysis['sub-category'], subcat_analysis['sales'], color="blue")
axes[0].set_title('Total Sales by Sub-Category')
axes[0].set_xlabel('Sub-Category')
axes[0].set_ylabel('Sales')
axes[0].tick_params(axis='x', rotation=90)

axes[1].bar(subcat_analysis['sub-category'], subcat_analysis['profit'], color="green")
axes[1].set_title('Total Profit by Sub-Category')
axes[1].set_xlabel('Sub-Category')
axes[1].set_ylabel('Profit')
axes[1].tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.show()
```

```python
monthly_sales = df.groupby(['year', 'month'])[['sales', 'profit']].sum().reset_index()
print(monthly_sales)
```

```python
df['date'] = pd.to_datetime(df[['year', 'month']].assign(day=1))
df = df.sort_values('date')

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Sales
axes[0].plot(df['date'], df['sales'], marker='o', label='Sales')
axes[0].set_title('Monthly Sales Trend')
axes[0].set_xlabel('Date')
axes[0].set_ylabel('Sales')
axes[0].legend()
axes[0].tick_params(axis='x', rotation=45)

# Profit
axes[1].plot(df['date'], df['profit'], marker='o', color='orange', label='Profit')
axes[1].set_title('Monthly Profit Trend')
axes[1].set_xlabel('Date')
axes[1].set_ylabel('Profit')
axes[1].legend()
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
